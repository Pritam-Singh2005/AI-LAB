```python
# ============================================================
# KAGGLE DATASET ANALYSIS IN GOOGLE COLAB
# Dataset:
# kristoft/pitt-quantum-repository-106066-molecules
# ============================================================


# ------------------------------------------------------------
# 1. INSTALL KAGGLEHUB
# ------------------------------------------------------------

!pip install -q kagglehub[pandas-datasets]


# ------------------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------------------

import kagglehub
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display


# ------------------------------------------------------------
# 3. DOWNLOAD THE DATASET
# ------------------------------------------------------------

dataset_path = kagglehub.dataset_download(
    "kristoft/pitt-quantum-repository-106066-molecules"
)

print("Dataset downloaded successfully!")
print("Dataset location:")
print(dataset_path)


# ------------------------------------------------------------
# 4. DISPLAY FILES PRESENT IN THE DATASET
# ------------------------------------------------------------

print("\nFiles available in the dataset:")

for root, directories, files in os.walk(dataset_path):
    for file in files:
        print(os.path.join(root, file))


# ------------------------------------------------------------
# 5. FIND JSON FILES (changed from CSV)
# ------------------------------------------------------------

json_files = [] # Changed variable name

for root, directories, files in os.walk(dataset_path):
    for file in files:
        if file.endswith(".json"): # Changed to .json
            json_files.append(os.path.join(root, file))

print("\nJSON files found:") # Changed print statement

for file in json_files:
    print(file)


# ------------------------------------------------------------
# 6. LOAD THE JSON DATASET (changed from CSV)
# ------------------------------------------------------------

if len(json_files) > 0: # Changed variable name

    file_path = json_files[0] # Changed variable name

    df = pd.read_json(file_path) # Changed to pd.read_json

    print("\nDataset loaded successfully!")

else:

    print("\nNo JSON file was found.") # Changed print statement
    df = None


# ------------------------------------------------------------
# 7. DISPLAY FIRST 5 RECORDS
# ------------------------------------------------------------

if df is not None:

    print("\nFirst 5 records:")
    print(df.head())


# ------------------------------------------------------------
# 8. DISPLAY LAST 5 RECORDS
# ------------------------------------------------------------

if df is not None:

    print("\nLast 5 records:")
    print(df.tail())


# ------------------------------------------------------------
# 9. CHECK DATASET SIZE
# ------------------------------------------------------------

if df is not None:

    print("\nDataset Shape:")
    print(df.shape)

    print("\nNumber of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])


# ------------------------------------------------------------
# 10. DISPLAY COLUMN NAMES
# ------------------------------------------------------------

if df is not None:

    print("\nColumn Names:")

    for column in df.columns:
        print(column)


# ------------------------------------------------------------
# 11. CHECK DATA TYPES AND INFORMATION
# ------------------------------------------------------------

if df is not None:

    print("\nDataset Information:")
    df.info()


# ------------------------------------------------------------
# 12. STATISTICAL SUMMARY
# ------------------------------------------------------------

if df is not None:

    print("\nStatistical Summary:")
    display(df.describe(include="all"))


# ------------------------------------------------------------
# 13. CHECK MISSING VALUES
# ------------------------------------------------------------

if df is not None:

    print("\nMissing Values:")
    print(df.isnull().sum())


# ------------------------------------------------------------
# 14. CHECK DUPLICATE RECORDS
# ------------------------------------------------------------

if df is not None:

    print("\nNumber of duplicate records:")

    print(df.duplicated().sum())


# ------------------------------------------------------------
# 15. REMOVE DUPLICATE RECORDS
# ------------------------------------------------------------

if df is not None:

    df = df.drop_duplicates()

    print("\nDataset shape after removing duplicates:")
    print(df.shape)


# ------------------------------------------------------------
# 16. DISPLAY NUMERICAL COLUMNS
# ------------------------------------------------------------

if df is not None:

    numerical_columns = df.select_dtypes(
        include=["int64", "float64"]
    ).columns

    print("\nNumerical Columns:")

    for column in numerical_columns:
        print(column)


# ------------------------------------------------------------
# 17. DISPLAY CATEGORICAL COLUMNS
# ------------------------------------------------------------

if df is not None:

    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns

    print("\nCategorical Columns:")

    for column in categorical_columns:
        print(column)


# ------------------------------------------------------------
# 18. VISUALIZE NUMERICAL FEATURES
# ------------------------------------------------------------

if df is not None:

    numerical_columns = df.select_dtypes(
        include=["int64", "float64"]
    ).columns

    for column in numerical_columns[:5]:

        plt.figure(figsize=(8, 5))

        plt.hist(df[column].dropna(), bins=30)

        plt.title("Distribution of " + column)

        plt.xlabel(column)

        plt.ylabel("Frequency")

        plt.show()


# ------------------------------------------------------------
# 19. CORRELATION HEATMAP
# ------------------------------------------------------------

if df is not None:

    numerical_df = df.select_dtypes(
        include=["int64", "float64"]
    )

    if numerical_df.shape[1] > 1:

        plt.figure(figsize=(12, 8))

        sns.heatmap(
            numerical_df.corr(),
            annot=True,
            cmap="coolwarm"
        )

        plt.title("Correlation Between Numerical Features")

        plt.show()


# ------------------------------------------------------------
# 20. FINAL DATASET INFORMATION
# ------------------------------------------------------------

if df is not None:

    print("\n======================================")
    print("FINAL DATASET SUMMARY")
    print("======================================")

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumns:")
    print(list(df.columns))

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate records:")
    print(df.duplicated().sum())
```
